```python
import os
import html
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Preformatted
from reportlab.lib.styles import getSampleStyleSheet


def replace_indents(code):
    # Replace each tab with four spaces
    return code.replace("    ", "  ")


def save_python_files_to_pdf(directory, output_pdf):
    doc = SimpleDocTemplate(output_pdf, pagesize=letter)
    styles = getSampleStyleSheet()
    flowables = []

    for root, _, files in os.walk(directory):
        for filename in files:
            if filename.endswith(".py"):
                file_path = os.path.join(root, filename)
                with open(file_path, "r") as file:
                    code = file.read()
                    # Replace tabs with spaces for indentation
                    code = replace_indents(code)
                    flowables.append(
                        Preformatted(
                            code,
                            styles["Code"],
                            newLineChars="> ",
                            maxLineLength=100,
                        )
                    )

    doc.build(flowables)


if __name__ == "__main__":
    # Provide the directory containing your Python files
    directory = "theoden"
    # Provide the output PDF file name
    output_pdf = "output.pdf"

    save_python_files_to_pdf(directory, output_pdf)
```
